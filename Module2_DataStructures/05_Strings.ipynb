### Slicing

Slicing is used to get multiple characters (a substring) from a string. Its syntax is similar to indexing, but instead of one index you use two numbers separated by a colon, e.g. `s[start:end]`.

- `s[start:end]` : characters from `start` through `end-1`
- `s[start:]` : characters from `start` through the end
- `s[:end]` : characters from the beginning through `end-1`
- `s[:]` : a copy of the whole string

One or both indices can be dropped. `monty_python[:5]` is equal to `monty_python[0:5]`.

````python
monty_python = "Monty Python"
monty = monty_python[:5]
print(monty)
````

**You should see:** `Monty`

### in operator

If you want to check whether a string contains a specific letter or a substring, you can use the `in` keyword. Run the code below.

````python
current_variable = "ice cream"
print("i" in current_variable)    # True if there is an i in the string, otherwise False
````

**You should see:** `True`

The `in` operator is quite useful combined with an `if-else` statement.

````python
#Try changing the value of my_string to a string that contains 'i'
my_string = "kare-kare"
if ("i" in my_string):
    print("I want an ice cream!")

else:
    print("I am on diet!")
````

With real data, `in` is a quick way to spot a *family* of values. All these 311 complaint types start with `Noise`, so we can test for a substring (note: `in` is case-sensitive).

````python
complaints = ['Noise - Street/Sidewalk', 'Illegal Parking', 'Noise - Commercial', 'Rodent']

for complaint in complaints:
    if "Noise" in complaint:
        print(complaint, "-> noise complaint")
    else:
        print(complaint, "-> something else")
````

**You should see:**

```text
Noise - Street/Sidewalk -> noise complaint
Illegal Parking -> something else
Noise - Commercial -> noise complaint
Rodent -> something else
```

### Comparing strings: `==`, not `is`

To ask "are these two strings equal?" use `==`. Avoid `is` for this. `is` asks whether two names point to the *very same object in memory*, which is not what you mean, and it can give surprising answers. Since Python 3.8, writing `x is "text"` even triggers a `SyntaxWarning`.

````python
borough = "QUEENS"
print(borough == "QUEENS")
print(borough != "BROOKLYN")
````

**You should see:**

```text
True
True
```

### Formatting Strings

To format strings we use `{ }` placeholders inside the quotes. Afterwards, we write `.format(`*value*`)`. A newer, shorter way is the **f-string**: put an `f` before the opening quote and write the variable straight into the braces.

Let's show both via an example.

````python
current_cgpa = 3.554

print("my current cgpa is {} ".format(current_cgpa))
print("my current cgpa in one decimal places is {:.1f}. Thank you".format(current_cgpa))
print("my current cgpa in one decimal places is {:20.1f}. Thank you".format(current_cgpa)) # We added 20 spacing (by default to the left)
print("my current cgpa in one decimal places is {:<20.1f}. Thank you".format(current_cgpa)) # < means the variable content is left-aligned.
print("my current cgpa in one decimal places is {:^20.1f}. Thank you".format(current_cgpa)) # ^ means the variable content is center-aligned.

# The same thing as an f-string:
print(f"my current cgpa in one decimal places is {current_cgpa:.1f}. Thank you")
````

Formatting is how analysts turn numbers into readable report lines. Here is a count of bicycles from the Montreal cycling dataset (Berri 1 bike path, 1 January 2012):

````python
day = "2012-01-01"
cyclists = 35
total_week = 933

print(f"{day}: {cyclists} cyclists ({cyclists / total_week:.1%} of the week)")
````

**You should see:** `2012-01-01: 35 cyclists (3.8% of the week)`

### String Length

The `len()` function is used to count how many characters a string contains.

````python
phrase = """
It is a really long string
triple-quoted strings are used
to define multi-line strings
"""

count = len(phrase)

print("Your character count is {} ".format(count))
````

Note: `len()` returns an output of type `int`. You can manipulate it like any integer.

````python
print ("Count added by 3 is {}".format(count + 3))
````

This is useful for dynamic slicing of your strings.

````python
# Recall: // means integer division
half = len(phrase) // 2
print ("The first half of your phrase is " + phrase[:half] )
````

### Built-in String Methods

There are a lot of useful string methods. The `lower()` method gets rid of capitalization; `upper()` makes a string uppercase. To call any string method, type a dot after the string (or a variable containing it) and then the method name, e.g. `"John".upper()`.

````python
my_name = "John Anthony Jose"

lower_name = my_name.lower()
print ("My Name in lower case " + lower_name)

upper_name = my_name.upper()
print("My Name in upper case " + upper_name)
````

**You should see:**

```text
My Name in lower case john anthony jose
My Name in upper case JOHN ANTHONY JOSE
```

#### Cleaning text: `strip`, `split`, `replace`, `startswith`

Real-world text is messy: stray spaces, inconsistent capitalization, several facts packed into one string. These four methods do most of the cleanup. Each one **returns a new string** (strings never change in place). Here we dissect a 311 complaint type, `'Noise - Street/Sidewalk'`, which holds a category and a detail separated by `' - '`.

````python
raw = "  Noise - Street/Sidewalk "

complaint = raw.strip()                 # remove spaces at both ends
print(repr(complaint))

category, detail = complaint.split(" - ")   # cut into pieces at ' - '
print(category)
print(detail)

print(complaint.lower())
print(complaint.replace("/", " or "))
print(complaint.startswith("Noise"))
````

**You should see:**

```text
'Noise - Street/Sidewalk'
Noise
Street/Sidewalk
noise - street/sidewalk
Noise - Street or Sidewalk
True
```

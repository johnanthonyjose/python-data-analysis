## Python as a calculator

We can use Python like a calculator. Consider the simple expression $3 + 8$. We can evaluate and print this by:

**You should see:** `11`

````python
3 + 8
````

Another simple calculation is the gravitational potential $V$ of a body of mass $m$ (point mass) at a distance $r$ from a body of mass $M$, which is given by

$$
V = \frac{G M m}{r}
$$

where $G$ is the *gravitational constant*. A good approximation is $G = 6.674 \times 10^{-11}$ N m$^{2}$ kg$^{-2}$.

For the case $M = 1.65 \times 10^{12}$ kg, $m = 6.1 \times 10^2$ kg and $r = 7.0 \times 10^3$ m, we can compute the  gravitational potential $V$:

**You should see:** a number close to `9.6` (about `9.596`).

````python
6.674e-11*1.65e12*6.1e2/7.0e3
````

We have used 'scientific notation' to input the values. For example, the number $8 \times 10^{-2}$ can be input as `0.08` or `8e-2`. We can easily verify that the two are the same via subtraction:

**You should see:** `0.0` (the two spellings of the number are equal).

````python
0.08 - 8e-2
````

## Putting it together: real numbers from the course data

Later in the course you will analyse two datasets: daily bicycle counts from Montreal (`bikes.csv`) and New York 311 service requests. Here are a few real values typed in by hand, so we can practise variables and expressions without any files.

Cyclists counted on the *Berri 1* bike path, Monday 2 to Friday 6 January 2012:

````python
monday = 83
tuesday = 135
wednesday = 144
thursday = 197
friday = 146

weekday_total = monday + tuesday + wednesday + thursday + friday
weekday_mean = weekday_total / 5
print("total:", weekday_total)
print("mean per day:", weekday_mean)
````

**You should see:**

```text
total: 705
mean per day: 141.0
```

And a share from the 311 data. Of the first 111,069 service requests, 14,200 were about `HEATING`. What percentage is that? (An f-string, `f"..."`, lets us put a number inside text, and `:.1f` rounds to one decimal.)

````python
heating = 14200
all_requests = 111069

share = heating / all_requests * 100
print(f"HEATING is {share:.1f}% of all requests")
````

**You should see:** `HEATING is 12.8% of all requests`
